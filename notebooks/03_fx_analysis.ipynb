# 03 - FX Transmission: CAD vs KRW
Compare how two differently exposed economies' currencies co-move with oil and US yields. This is descriptive analysis, not a causal model.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(ROOT))
import pandas as pd
import statsmodels.api as sm
from src.data_loader import load_sample_levels
from src.indicators import to_moves, rolling_correlation
```

```python
levels = load_sample_levels().loc['2015':, ['WTI', 'US10Y', 'USDCAD', 'USDKRW', 'SPY']]
moves = to_moves(levels).dropna()
moves.describe().T
```

```python
results = {}
for currency in ['USDCAD', 'USDKRW']:
    y = moves[currency] * 100
    X = sm.add_constant(pd.DataFrame({'WTI_return_pct': moves['WTI'] * 100, 'US10Y_change_bps': moves['US10Y'], 'SPY_return_pct': moves['SPY'] * 100}))
    results[currency] = sm.OLS(y, X).fit(cov_type='HAC', cov_kwds={'maxlags': 5})
    print(currency, results[currency].params.round(4).to_dict())
```
